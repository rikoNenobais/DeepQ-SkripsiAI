```python
import pickle
import pandas as pd

path = r"C:\Users\HP\Downloads\Deep-Q-learning-TF2-modernized\results\results_final.pkl"

with open(path, "rb") as f:
    data = pickle.load(f)

rows = []

for iot in range(len(data["reward"])):

    rewards = data["reward"][iot]
    delays = data["delay"][iot]
    actions = data["action"][iot]

    for episode in range(len(rewards)):

        for time in range(len(rewards[episode])):

            rows.append({
                "episode": episode,
                "iot": iot,
                "time": time,
                "reward": rewards[episode][time],
                "delay": delays[episode][time],
                "action": actions[episode][time]
            })

df = pd.DataFrame(rows)

df.to_csv("result_final.csv", index=False)
print(df.shape)
```
